```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv(r'C:\Users\rupab\OneDrive\Desktop\ML_PRACTICE\LOGISTIC_REGRESSION\DATASETS\logistic_regression_dataset.csv')
```

```python
df.head(5)
```

```python
X = df.drop(['Purchased'],axis = 1)
Y = df['Purchased']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,Y,test_size = 0.2,random_state=42)
```

```python
model = LogisticRegression()
scale = StandardScaler()
```

```python
model.fit(X_test,y_test)
y_pred = model.predict(X_test)
```

```python
print('Accuracy_score',accuracy_score(y_test,y_pred))
```

```python
X_train_scaled = scale.fit_transform(X_train)
X_test_scaled = scale.transform(X_test)
```

```python
model.fit(X_train_scaled,y_train)
```

```python
y_scaled_pred = model.predict(X_test_scaled)
```

```python
print('Scaled_Accuracy',accuracy_score(y_test,y_scaled_pred))
```
